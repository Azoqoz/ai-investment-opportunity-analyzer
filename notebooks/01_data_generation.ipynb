```python
# AI Investment Opportunity Analyzer
# Notebook 01: Synthetic Data Generation

import pandas as pd
import numpy as np
from pathlib import Path

# Reproducibility
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

# Project paths
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_PATH = PROJECT_ROOT / "data" / "processed"

RAW_DATA_PATH.mkdir(parents=True, exist_ok=True)
PROCESSED_DATA_PATH.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Raw data path:", RAW_DATA_PATH)
print("Processed data path:", PROCESSED_DATA_PATH)
```

```python
# Dataset configuration

N_SAMPLES = 5000

sectors = [
    "Real Estate",
    "Tourism",
    "Entertainment",
    "Healthcare",
    "Logistics",
    "Renewable Energy",
    "Technology",
    "Education",
    "Retail",
    "Infrastructure"
]

regions = [
    "Riyadh",
    "Makkah",
    "Madinah",
    "Eastern Province",
    "Qassim",
    "Asir",
    "Tabuk",
    "Jazan"
]

competition_levels = [
    "Low",
    "Medium",
    "High"
]

print("Number of samples:", N_SAMPLES)
print("Number of sectors:", len(sectors))
print("Number of regions:", len(regions))
print("Competition levels:", competition_levels)
```

```python
# Generate base opportunity information

opportunity_ids = [f"INV-{i:05d}" for i in range(1, N_SAMPLES + 1)]

selected_sectors = np.random.choice(sectors, size=N_SAMPLES)
selected_regions = np.random.choice(regions, size=N_SAMPLES)
selected_competition = np.random.choice(
    competition_levels,
    size=N_SAMPLES,
    p=[0.30, 0.50, 0.20]
)

opportunity_names = [
    f"{region} {sector} Opportunity"
    for region, sector in zip(selected_regions, selected_sectors)
]

df = pd.DataFrame({
    "opportunity_id": opportunity_ids,
    "opportunity_name": opportunity_names,
    "sector": selected_sectors,
    "region": selected_regions,
    "competition_level": selected_competition
})

df.head()
```

```python
df.shape
```

```python
# Generate financial features

# Investment size in million SAR
df["investment_size_million"] = np.round(
    np.random.lognormal(mean=5.2, sigma=1.0, size=N_SAMPLES),
    2
)

# Limit extreme values to keep data realistic
df["investment_size_million"] = df["investment_size_million"].clip(20, 5000)

# CAPEX usually represents a large part of investment size
df["capex_million"] = np.round(
    df["investment_size_million"] * np.random.uniform(0.55, 0.90, size=N_SAMPLES),
    2
)

# OPEX is annual operating cost
df["opex_million"] = np.round(
    df["investment_size_million"] * np.random.uniform(0.05, 0.25, size=N_SAMPLES),
    2
)

# Expected ROI varies by opportunity
df["expected_roi_percent"] = np.round(
    np.random.normal(loc=11, scale=6, size=N_SAMPLES),
    2
).clip(-5, 30)

# IRR is related to ROI but with some variation
df["irr_percent"] = np.round(
    df["expected_roi_percent"] + np.random.normal(loc=2, scale=4, size=N_SAMPLES),
    2
).clip(-5, 35)

# Payback period: lower is better
df["payback_period_years"] = np.round(
    np.random.normal(loc=7, scale=2.5, size=N_SAMPLES),
    2
).clip(1, 15)

# Revenue growth
df["revenue_growth_percent"] = np.round(
    np.random.normal(loc=8, scale=5, size=N_SAMPLES),
    2
).clip(-5, 25)

# Profit margin
df["profit_margin_percent"] = np.round(
    np.random.normal(loc=18, scale=9, size=N_SAMPLES),
    2
).clip(-10, 40)

# NPV depends partly on investment size and expected ROI
df["npv_million"] = np.round(
    (df["investment_size_million"] * (df["expected_roi_percent"] / 100) * np.random.uniform(2, 8, size=N_SAMPLES))
    - (df["investment_size_million"] * np.random.uniform(0.1, 0.5, size=N_SAMPLES)),
    2
).clip(-500, 3000)

df.head()
```

```python
df[
    [
        "investment_size_million",
        "expected_roi_percent",
        "irr_percent",
        "npv_million",
        "payback_period_years",
        "profit_margin_percent"
    ]
].describe()
```

```python
# Generate market features

# Market size in billion SAR
df["market_size_billion"] = np.round(
    np.random.lognormal(mean=3.2, sigma=0.8, size=N_SAMPLES),
    2
).clip(1, 300)

# Market growth percentage
df["market_growth_percent"] = np.round(
    np.random.normal(loc=7, scale=5, size=N_SAMPLES),
    2
).clip(-3, 20)

# Demand score from 0 to 100
df["demand_score"] = np.round(
    np.random.normal(loc=65, scale=18, size=N_SAMPLES),
    2
).clip(0, 100)

# Customer adoption score from 0 to 100
df["customer_adoption_score"] = np.round(
    np.random.normal(loc=60, scale=20, size=N_SAMPLES),
    2
).clip(0, 100)

# Scalability score from 0 to 100
df["scalability_score"] = np.round(
    np.random.normal(loc=62, scale=19, size=N_SAMPLES),
    2
).clip(0, 100)

df.head()
```

```python
df[
    [
        "market_size_billion",
        "market_growth_percent",
        "demand_score",
        "customer_adoption_score",
        "scalability_score"
    ]
].describe()
```

```python
# Generate risk features

# Base risk scores from 0 to 100
df["financial_risk_score"] = np.round(
    np.random.normal(loc=45, scale=20, size=N_SAMPLES),
    2
).clip(0, 100)

df["regulatory_risk_score"] = np.round(
    np.random.normal(loc=40, scale=22, size=N_SAMPLES),
    2
).clip(0, 100)

df["execution_risk_score"] = np.round(
    np.random.normal(loc=50, scale=20, size=N_SAMPLES),
    2
).clip(0, 100)

df["market_risk_score"] = np.round(
    np.random.normal(loc=43, scale=21, size=N_SAMPLES),
    2
).clip(0, 100)

df["dependency_risk_score"] = np.round(
    np.random.normal(loc=38, scale=22, size=N_SAMPLES),
    2
).clip(0, 100)

# Competition effect: high competition increases market risk slightly
competition_risk_adjustment = df["competition_level"].map({
    "Low": -5,
    "Medium": 0,
    "High": 8
})

df["market_risk_score"] = (
    df["market_risk_score"] + competition_risk_adjustment
).clip(0, 100)

# Overall risk score
df["overall_risk_score"] = np.round(
    df[
        [
            "financial_risk_score",
            "regulatory_risk_score",
            "execution_risk_score",
            "market_risk_score",
            "dependency_risk_score"
        ]
    ].mean(axis=1),
    2
)

df.head()
```

```python
df[
    [
        "financial_risk_score",
        "regulatory_risk_score",
        "execution_risk_score",
        "market_risk_score",
        "dependency_risk_score",
        "overall_risk_score"
    ]
].describe()
```

```python
# Generate strategic alignment features

# Scores from 0 to 100
df["vision_2030_alignment_score"] = np.round(
    np.random.normal(loc=68, scale=18, size=N_SAMPLES),
    2
).clip(0, 100)

df["economic_diversification_score"] = np.round(
    np.random.normal(loc=64, scale=19, size=N_SAMPLES),
    2
).clip(0, 100)

df["job_creation_score"] = np.round(
    np.random.normal(loc=58, scale=22, size=N_SAMPLES),
    2
).clip(0, 100)

df["localization_score"] = np.round(
    np.random.normal(loc=60, scale=20, size=N_SAMPLES),
    2
).clip(0, 100)

df["quality_of_life_score"] = np.round(
    np.random.normal(loc=62, scale=21, size=N_SAMPLES),
    2
).clip(0, 100)

# Sector-specific strategic boost
sector_strategic_boost = df["sector"].map({
    "Real Estate": 8,
    "Tourism": 10,
    "Entertainment": 8,
    "Healthcare": 9,
    "Logistics": 7,
    "Renewable Energy": 12,
    "Technology": 10,
    "Education": 8,
    "Retail": 3,
    "Infrastructure": 9
})

strategic_columns = [
    "vision_2030_alignment_score",
    "economic_diversification_score",
    "job_creation_score",
    "localization_score",
    "quality_of_life_score"
]

for col in strategic_columns:
    df[col] = (df[col] + sector_strategic_boost).clip(0, 100)

# Overall strategic impact score
df["strategic_impact_score"] = np.round(
    df[strategic_columns].mean(axis=1),
    2
)

df.head()
```

```python
# Generate ESG / sustainability features

df["environmental_impact_score"] = np.round(
    np.random.normal(loc=58, scale=22, size=N_SAMPLES),
    2
).clip(0, 100)

df["social_impact_score"] = np.round(
    np.random.normal(loc=62, scale=20, size=N_SAMPLES),
    2
).clip(0, 100)

df["governance_score"] = np.round(
    np.random.normal(loc=66, scale=18, size=N_SAMPLES),
    2
).clip(0, 100)

# Sector-specific sustainability adjustment
sector_sustainability_boost = df["sector"].map({
    "Real Estate": 4,
    "Tourism": 5,
    "Entertainment": 3,
    "Healthcare": 6,
    "Logistics": 2,
    "Renewable Energy": 14,
    "Technology": 5,
    "Education": 6,
    "Retail": 1,
    "Infrastructure": 4
})

esg_columns = [
    "environmental_impact_score",
    "social_impact_score",
    "governance_score"
]

for col in esg_columns:
    df[col] = (df[col] + sector_sustainability_boost).clip(0, 100)

df["sustainability_score"] = np.round(
    df[esg_columns].mean(axis=1),
    2
)

df.head()
```

```python
df[
    [
        "environmental_impact_score",
        "social_impact_score",
        "governance_score",
        "sustainability_score"
    ]
].describe()
```

```python
# Feature engineering

# Risk-adjusted ROI:
# Higher ROI is good, but high risk should reduce attractiveness.
df["risk_adjusted_roi"] = np.round(
    df["expected_roi_percent"] / (df["overall_risk_score"] + 1),
    4
)

# Profitability index:
# Measures NPV relative to investment size.
df["profitability_index"] = np.round(
    df["npv_million"] / (df["investment_size_million"] + 1),
    4
)

# Payback efficiency:
# Higher ROI with shorter payback period is better.
df["payback_efficiency"] = np.round(
    df["expected_roi_percent"] / (df["payback_period_years"] + 1),
    4
)

# Market attractiveness:
# Combines market growth, demand, customer adoption, and scalability.
df["market_attractiveness_score"] = np.round(
    (
        (df["market_growth_percent"].clip(0, 20) / 20) * 25
        + (df["demand_score"] / 100) * 25
        + (df["customer_adoption_score"] / 100) * 25
        + (df["scalability_score"] / 100) * 25
    ),
    2
)

# Financial strength:
# Combines ROI, IRR, NPV, and profit margin into one normalized score.
df["financial_strength_score"] = np.round(
    (
        ((df["expected_roi_percent"] + 5) / 35) * 25
        + ((df["irr_percent"] + 5) / 40) * 25
        + ((df["npv_million"] + 500) / 3500) * 25
        + ((df["profit_margin_percent"] + 10) / 50) * 25
    ),
    2
).clip(0, 100)

df.head()
```

```python
df[
    [
        "risk_adjusted_roi",
        "profitability_index",
        "payback_efficiency",
        "market_attractiveness_score",
        "financial_strength_score"
    ]
].describe()
```

```python
# Generate target variable: investment_score

# Competition penalty
competition_penalty = df["competition_level"].map({
    "Low": 4,
    "Medium": 0,
    "High": -6
})

# Payback score: shorter payback is better
payback_score = np.round(
    (1 - ((df["payback_period_years"] - 1) / 14)) * 100,
    2
).clip(0, 100)

# Risk score inverted: lower risk is better
risk_score_inverted = 100 - df["overall_risk_score"]

# Base investment score before calibration
base_score = (
    df["financial_strength_score"] * 0.32
    + df["market_attractiveness_score"] * 0.18
    + df["strategic_impact_score"] * 0.20
    + df["sustainability_score"] * 0.10
    + risk_score_inverted * 0.15
    + payback_score * 0.05
    + competition_penalty
)

# Add realistic noise so the target is not perfectly rule-based
noise = np.random.normal(loc=0, scale=2.5, size=N_SAMPLES)

raw_score = base_score + noise

# Calibrate score to use a wider 0–100 range
# This keeps relative ranking but improves score distribution.
lower_bound = np.percentile(raw_score, 2)
upper_bound = np.percentile(raw_score, 98)

calibrated_score = (
    (raw_score - lower_bound) / (upper_bound - lower_bound)
) * 100

df["investment_score"] = np.round(
    calibrated_score.clip(0, 100),
    2
)

# Convert score into recommendation
def assign_recommendation(score):
    if score >= 75:
        return "Invest"
    elif score >= 45:
        return "Review"
    else:
        return "Reject"

df["recommendation"] = df["investment_score"].apply(assign_recommendation)

df[
    [
        "investment_score",
        "recommendation",
        "financial_strength_score",
        "market_attractiveness_score",
        "strategic_impact_score",
        "sustainability_score",
        "overall_risk_score"
    ]
].head()
```
